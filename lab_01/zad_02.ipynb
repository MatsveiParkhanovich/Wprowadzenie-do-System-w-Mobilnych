```python
import math

lam = 254
mu = 51
servers = 5

traffic = lam / mu
rho = lam / (servers * mu)

if rho >= 1:
    print("System jest niestabilny.")
else:
    sum_terms = sum(
        traffic ** n / math.factorial(n)
        for n in range(servers)
    )

    last_term = (
        traffic ** servers
        / math.factorial(servers)
        / (1 - rho)
    )

    p0 = 1 / (sum_terms + last_term)
    p_wait = last_term * p0
    p_no_wait = 1 - p_wait

    p_queue_more_than_two = (
        p0
        * traffic ** servers
        / math.factorial(servers)
        * rho ** 3
        / (1 - rho)
    )

    p_wait_over_15 = p_wait * math.exp(
        -(servers * mu - lam) * 15
    )

    lq = p_wait * rho / (1 - rho)
    wq = lq / lam

    print("ANALIZA SYSTEMU OBSLUGI")
    print(f"Lambda: {lam}")
    print(f"Mu: {mu}")
    print(f"Liczba serwerow: {servers}")
    print(f"Oferowana intensywnosc ruchu A: {traffic:.6f} Erlanga")
    print(f"Wspolczynnik obciazenia rho: {rho:.6f}")
    print()

    print("a. Intensywnosc ruchu")
    print(f"A = {traffic:.6f} Erlanga")
    print(f"rho = {rho:.6f}")
    print()

    print("b. Stabilnosc systemu")
    print(f"System stabilny: {'Tak' if rho < 1 else 'Nie'}")
    print(f"Minimalna liczba serwerow: {math.floor(lam / mu) + 1}")
    print()

    print("c. Prawdopodobienstwo, ze nie bedzie kolejki")
    print(f"P = {p_no_wait:.8f}")
    print()

    print("d. Prawdopodobienstwo, ze klient bedzie czekac")
    print(f"P = {p_wait:.8f}")
    print()

    print("e. Prawdopodobienstwo, ze w kolejce czekaja wiecej niz 2 osoby")
    print(f"P = {p_queue_more_than_two:.8f}")
    print()

    print("f. Prawdopodobienstwo oczekiwania dluzej niz 15 minut")
    print(f"P = {p_wait_over_15:.10f}")
    print()

    print("g. Srednia liczba klientow oczekujacych w kolejce")
    print(f"Lq = {lq:.6f}")
    print()

    print("h. Sredni czas oczekiwania na obsluge")
    print(f"Wq = {wq:.6f} min")
    print(f"Wq = {wq * 60:.4f} sek")
    print()

    print("Kontrola zgodnosci wynikow")
    print(f"lambda * Wq = {lam * wq:.6f}")
    print(f"Lq = {lq:.6f}")
```

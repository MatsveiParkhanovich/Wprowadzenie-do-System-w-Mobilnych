```python
lambda_rate = 29
mu_rate = 7

traffic_load = lambda_rate / mu_rate

print("Analiza sieci komórkowej")
print(f"Intensywność zgłoszeń lambda: {lambda_rate} na minutę")
print(f"Intensywność obsługi mu: {mu_rate} na minutę")
print(f"Oferowana intensywność ruchu: {traffic_load:.4f} Erlanga")
print()

print("Analiza w zależności od liczby kanałów:")

for r in range(1, 11):
    rho = lambda_rate / (r * mu_rate)
    equilibrium = lambda_rate < r * mu_rate

    print(
        f"Liczba kanałów: {r}, "
        f"rho: {rho:.4f}, "
        f"Stan równowagi: {'Tak' if equilibrium else 'Nie'}"
    )

minimum_channels = int(lambda_rate // mu_rate) + 1

print()
print(f"Minimalna liczba kanałów zapewniająca równowagę: {minimum_channels}")
```

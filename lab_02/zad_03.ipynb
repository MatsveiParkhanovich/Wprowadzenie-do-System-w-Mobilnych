```python
intervals = [0, 8, 17, 9, 11, 12, 13, 5, 4, 11, 5, 8, 7, 5, 8, 7, 5, 4, 16, 5]

service_time = 40 / 3

arrival_times = []
current_time = 0

for i, interval in enumerate(intervals):
    if i > 0:
        current_time += interval
    arrival_times.append(current_time)


def simulate_system(arrivals, service_time, number_of_cabinets):
    free_times = [0.0] * number_of_cabinets
    start_times = []
    waiting_times = []
    queue_lengths = []

    for arrival in arrivals:
        queue_lengths.append(sum(start > arrival for start in start_times))

        cabinet = min(range(number_of_cabinets), key=lambda i: free_times[i])
        start_time = max(arrival, free_times[cabinet])

        waiting_times.append(start_time - arrival)
        start_times.append(start_time)
        free_times[cabinet] = start_time + service_time

    probability_no_wait = sum(wait == 0 for wait in waiting_times) / len(arrivals)
    average_queue = sum(queue_lengths) / len(queue_lengths)
    average_wait = sum(waiting_times) / len(waiting_times)

    return probability_no_wait, average_queue, average_wait


print("ZADANIE 3 - ANALIZA PRZYCHODNI")
print(f"Czas obslugi pacjenta: {service_time:.2f} min")
print()

for cabinets in [3, 2]:
    probability, average_queue, average_wait = simulate_system(
        arrival_times, service_time, cabinets
    )

    print(f"Liczba gabinetow: {cabinets}")
    print(f"Prawdopodobienstwo braku oczekiwania: {probability:.4f}")
    print(f"Srednia liczba oczekujacych przy przybyciu: {average_queue:.4f}")
    print(f"Sredni czas oczekiwania: {average_wait:.4f} min")
    print()
```

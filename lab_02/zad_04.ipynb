```python
arrival_intervals = [
    0, 6, 40, 30, 50, 30, 25, 5,
    15, 15, 10, 30, 40, 40, 40, 50,
    50, 55, 45, 55, 55, 50, 45, 45,
    50, 10, 25, 25, 55, 55, 40, 50
]

service_times = [
    40, 60, 65, 60, 40, 55, 65, 60,
    60, 70, 65, 70, 65, 75, 60, 50,
    65, 65, 65, 75, 70, 60, 65, 62,
    60, 40, 55, 75, 60, 40, 70, 60
]

servers = 2
number_of_students = len(arrival_intervals)

total_arrival_time = sum(arrival_intervals)
total_service_time = sum(service_times)

lambda_rate = number_of_students / total_arrival_time
mu_rate = number_of_students / total_service_time
rho = lambda_rate / (servers * mu_rate)

print("ZADANIE 4 - ANALIZA BARY")
print(f"Liczba studentow: {number_of_students}")
print(f"Liczba stanowisk obslugi: {servers}")
print(f"Calkowity czas obserwacji: {total_arrival_time} s")
print(f"Suma czasow obslugi: {total_service_time} s")
print()

print("a. Srednie intensywnosci")
print(f"Lambda: {lambda_rate:.6f} studentow/s")
print(f"Lambda: {lambda_rate * 60:.4f} studentow/min")
print(f"Mu jednego stanowiska: {mu_rate:.6f} studentow/s")
print(f"Mu jednego stanowiska: {mu_rate * 60:.4f} studentow/min")
print()

print("b. Diagnoza systemu")
print(f"Rho: {rho:.6f}")
print(f"Wykorzystanie stanowisk: {rho * 100:.2f}%")

if rho < 1:
    print("System jest stabilny wedlug modelu kolejkowego.")
elif rho == 1:
    print("System pracuje na granicy stabilnosci.")
else:
    print("System jest niestabilny wedlug modelu kolejkowego.")
```

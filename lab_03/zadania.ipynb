## Zadanie 2 – Odległość wykorzystania i interferencja współkanałowa

**Treść zadania:**

Klaster sieci komórkowej składa się z 7 komórek. Ile wynosi odległość wykorzystania (reuse distance)? Określ numery komórek klastra. Ile wynosi wartość interferencji współkanałowej (cochannel interference) w najgorszym przypadku położenia stacji mobilnej?

### 1. Obliczenie odległości wykorzystania (Reuse Distance)

Wielkość klastra wynosi:

$$
N=7
$$

Korzystamy ze wzoru:

$$
N=i^2+ij+j^2
$$

Dla klastra składającego się z 7 komórek przyjmujemy:

$$
i=2,\quad j=1
$$

Sprawdzenie:

$$
N=2^2+2\cdot1+1^2=7
$$

Odległość wykorzystania obliczamy ze wzoru:

$$
D=\sqrt{3N}\cdot R
$$

Po podstawieniu:

$$
D=\sqrt{3\cdot7}\cdot R
$$

$$
\boxed{D=\sqrt{21}\cdot R\approx4.583R}
$$

Oznacza to, że odległość między komórkami wykorzystującymi te same częstotliwości wynosi około 4.583 promienia komórki.

### 2. Numerowanie komórek klastra

Korzystamy ze wzoru:

$$
L=[(i+1)u+v]\mod N
$$

Po podstawieniu parametrów:

$$
L=(3u+v)\mod7
$$

Przyjmujemy środkową komórkę o współrzędnych (0,0), a następnie sześć sąsiednich komórek.

| Komórka | u | v | Numer L |
|---|---:|---:|---:|
| Środkowa | 0 | 0 | 0 |
| 1 | 1 | 0 | 3 |
| 2 | 0 | 1 | 1 |
| 3 | -1 | 1 | 5 |
| 4 | -1 | 0 | 4 |
| 5 | 0 | -1 | 6 |
| 6 | 1 | -1 | 2 |

Otrzymane numery komórek klastra to:

$$
\boxed{0,1,2,3,4,5,6}
$$

Każda komórka otrzymuje inny numer.

### 3. Interferencja współkanałowa

Interferencja współkanałowa występuje pomiędzy komórkami wykorzystującymi te same częstotliwości.

Korzystamy ze wzoru podanego w instrukcji:

$$
I=\sum_{k=1}^{M}\left(\frac{D_k}{R}\right)^\gamma
$$

gdzie:

- $M$ – liczba stacji bazowych powodujących interferencję,
- $D_k$ – odległość pomiędzy stacją bazową a stacją mobilną,
- $R$ – promień komórki,
- $\gamma$ – wykładnik propagacji, należący do przedziału $(2,5)$.

W klastrze składającym się z 7 komórek przy standardowym założeniu pierwszego pierścienia interferencyjnego występuje 6 stacji bazowych wykorzystujących te same częstotliwości.

W najgorszym przypadku należy uwzględnić położenie stacji mobilnej, które powoduje największy wpływ stacji zakłócających.

**Uwaga:** Instrukcja nie podaje konkretnej wartości wykładnika $\gamma$, promienia komórki ani położenia stacji mobilnej. Dlatego nie można wyznaczyć jednej liczbowej wartości interferencji bez dodatkowych założeń.

### Wnioski

Dla klastra składającego się z 7 komórek otrzymano parametry $i=2$ oraz $j=1$. Odległość wykorzystania wynosi około 4.583 promienia komórki. Wszystkie komórki klastra otrzymały różne numery od 0 do 6.

Interferencja współkanałowa zależy od odległości między stacją mobilną a stacjami bazowymi wykorzystującymi te same częstotliwości oraz od warunków propagacji fal radiowych.

## Zadanie 4 – Architektura sieci komórkowej

### Treść zadania

Rysunek przedstawia architekturę sieci komórkowej. Jakie potencjalne czynniki spowodowały konieczność budowy sieci o takiej architekturze?

### Odpowiedź

Konieczność zastosowania architektury komórkowej wynika z kilku podstawowych czynników:

- **Ograniczona liczba częstotliwości** – dostępne pasmo radiowe jest ograniczone, dlatego konieczne jest ponowne wykorzystywanie tych samych częstotliwości w różnych komórkach.
- **Rosnąca liczba użytkowników** – zwiększająca się liczba urządzeń mobilnych wymaga większej pojemności sieci i możliwości obsługi wielu połączeń jednocześnie.
- **Zasięg sieci** – podział obszaru na mniejsze komórki pozwala zapewnić dostęp do usług na dużym obszarze.
- **Zakłócenia radiowe** – odpowiednie rozmieszczenie stacji bazowych i przydzielanie częstotliwości pozwala ograniczyć interferencję współkanałową.
- **Mobilność użytkowników** – architektura komórkowa umożliwia przekazywanie aktywnych połączeń między sąsiednimi stacjami bazowymi (handover).

### Wnioski

Architektura komórkowa pozwala efektywnie wykorzystać ograniczone zasoby częstotliwościowe, zwiększyć pojemność sieci, zapewnić odpowiedni zasięg oraz umożliwić swobodne przemieszczanie się użytkowników bez przerywania połączeń.

## Zadanie 6 – Podział klastra na mniejsze klastry

### Treść zadania

W pewnej sieci każdy klaster składa się z 16 komórek. Czy jest możliwa zamiana każdego klastra na dwa mniejsze po 7 i 9 komórek?

### Odpowiedź

Nie jest możliwa taka zamiana w standardowym, jednolitym modelu planowania sieci komórkowej.

Rozmiar klastra musi spełniać zależność:

$$
N = i^2 + i \cdot j + j^2
$$

gdzie:
- $N$ – liczba komórek w klastrze,
- $i, j$ – parametry określające rozmieszczenie komórek.

Wartości 7 i 9 są dopuszczalnymi rozmiarami klastrów:

- Dla $N = 7$: $i = 2$, $j = 1$.

$$
N = 2^2 + 2 \cdot 1 + 1^2 = 7
$$

- Dla $N = 9$: $i = 3$, $j = 0$.

$$
N = 3^2 + 3 \cdot 0 + 0^2 = 9
$$

Jednak samo dodanie tych wartości nie gwarantuje poprawnego podziału klastra. Klastry o różnych rozmiarach mają różne odległości ponownego wykorzystania częstotliwości, co może prowadzić do zwiększenia interferencji współkanałowej.

### Wnioski

Mimo że klastry o rozmiarach 7 i 9 są matematycznie dopuszczalne, nie można po prostu zastąpić nimi każdego klastra o rozmiarze 16. Wymagałoby to ponownego zaplanowania rozmieszczenia stacji bazowych i przydziału częstotliwości, aby zapewnić odpowiedni poziom ochrony przed zakłóceniami.

## Zadanie 7 – Odległość wykorzystania i pojemność sieci

### Treść zadania

Dla przedstawionego schematu sieci komórkowej należy obliczyć odległość ponownego wykorzystania częstotliwości oraz liczbę abonentów, którzy mogą jednocześnie prowadzić rozmowy.

### Rozwiązanie

**1. Obliczenie odległości ponownego wykorzystania częstotliwości**

Ze schematu wynika, że klaster składa się z 12 komórek.

Wykorzystujemy wzór:

$$
N=i^2+i\cdot j+j^2
$$

Dla parametrów \(i=2\) oraz \(j=2\):

$$
N=2^2+2\cdot2+2^2=12
$$

Odległość ponownego wykorzystania częstotliwości obliczamy ze wzoru:

$$
D=R\sqrt{3N}
$$

Średnica komórki wynosi 2 km, dlatego:

$$
R=\frac{2}{2}=1\text{ km}
$$

Podstawiamy wartości:

$$
D=1\cdot\sqrt{3\cdot12}
$$

$$
\boxed{D=6\text{ km}}
$$

**2. Obliczenie liczby dostępnych kanałów**

Całkowite pasmo wynosi 30 MHz, a szerokość jednego kanału wynosi 25 kHz.

Liczbę dostępnych kanałów obliczamy:

$$
K=\frac{30\,000\,000}{25\,000}=1200
$$

Ponieważ klaster składa się z 12 komórek, liczba kanałów przypadających na jedną komórkę wynosi:

$$
K_{\text{kom}}=\frac{1200}{12}=100
$$

Dla każdej komórki rezerwujemy 10 kanałów do celów kontrolnych:

$$
K_{\text{rozm}}=100-10=90
$$

**3. Obliczenie liczby abonentów**

Kodowanie TDMA umożliwia obsługę 8 abonentów na jednym kanale.

Liczbę abonentów mogących jednocześnie rozmawiać w jednej komórce obliczamy:

$$
A=90\cdot8
$$

$$
\boxed{A=720\text{ abonentów}}
$$

Dla całego klastra składającego się z 12 komórek:

$$
A_{\text{klaster}}=720\cdot12
$$

$$
\boxed{A_{\text{klaster}}=8640\text{ abonentów}}
$$

### Wnioski

Odległość ponownego wykorzystania częstotliwości wynosi 6 km. W każdej komórce dostępnych jest 90 kanałów przeznaczonych do obsługi rozmów, co przy zastosowaniu technologii TDMA pozwala na jednoczesną obsługę 720 abonentów.

Cały klaster składający się z 12 komórek może obsługiwać jednocześnie 8640 abonentów.

# Zadanie 9 – Przeniesienie połączenia (Handoff)

## Treść zadania

Co oznacza przeniesienie połączenia (handoff) i obszar przeniesienia (handoff region)? Wyjaśnij pojęcia przy pomocy odpowiedniego rysunku.

## Odpowiedź

### 1. Handoff – przeniesienie połączenia

Handoff to proces automatycznego przekazywania aktywnego połączenia użytkownika z jednej stacji bazowej do drugiej podczas przemieszczania się między komórkami sieci.

Proces ten pozwala zachować ciągłość rozmowy bez konieczności jej przerywania.

Przeniesienie połączenia następuje zazwyczaj wtedy, gdy sygnał aktualnej stacji bazowej słabnie, a sygnał sąsiedniej stacji staje się odpowiednio silny.

### 2. Handoff region – obszar przeniesienia

Handoff region to obszar znajdujący się w pobliżu granicy sąsiednich komórek, w którym może nastąpić przekazanie połączenia między stacjami bazowymi.

W tym obszarze telefon może odbierać sygnały z obu stacji bazowych. Sieć analizuje ich jakość i wybiera odpowiedni moment na przeniesienie połączenia.

### 3. Przebieg procesu handoff

1. Telefon znajduje się w zasięgu pierwszej stacji bazowej.
2. Użytkownik przemieszcza się w kierunku sąsiedniej komórki.
3. Sygnał pierwszej stacji bazowej stopniowo słabnie.
4. Telefon znajduje się w obszarze handoff.
5. Połączenie zostaje przekazane do drugiej stacji bazowej.

## Wnioski

Handoff umożliwia zachowanie ciągłości połączenia podczas przemieszczania się użytkownika między komórkami sieci. Handoff region jest obszarem, w którym może nastąpić przekazanie połączenia między sąsiednimi stacjami bazowymi.

Dzięki temu rozwiązaniu użytkownik może swobodnie przemieszczać się podczas rozmowy bez jej przerywania.


```python
import matplotlib.pyplot as plt
from matplotlib.patches import Circle

fig, ax = plt.subplots(figsize=(10, 5))

# Obszary dwóch sąsiednich stacji bazowych
cell1 = Circle((0, 0), 3, color='skyblue', alpha=0.25)
cell2 = Circle((4, 0), 3, color='lightgreen', alpha=0.25)

ax.add_patch(cell1)
ax.add_patch(cell2)

# Stacje bazowe
ax.scatter(0, 0, color='blue', s=120, zorder=5)
ax.scatter(4, 0, color='green', s=120, zorder=5)

ax.text(0, -0.4, 'Stacja bazowa 1', ha='center')
ax.text(4, -0.4, 'Stacja bazowa 2', ha='center')

# Obszar handoff
handoff = Circle((2, 0), 1, color='orange', alpha=0.35)
ax.add_patch(handoff)

ax.text(2, 0.3, 'Handoff region', ha='center',
        fontsize=11, fontweight='bold')

# Kierunek ruchu użytkownika
ax.annotate(
    'Ruch użytkownika',
    xy=(4.8, 1.2),
    xytext=(-0.8, 1.2),
    arrowprops=dict(arrowstyle='->', linewidth=2),
    fontsize=11
)

ax.set_xlim(-3.5, 7.5)
ax.set_ylim(-3.5, 3.5)
ax.set_aspect('equal')
ax.set_title('Proces przekazywania połączenia (Handoff)')
ax.axis('off')

plt.show()
```

## Zadanie 10 – Interferencja sąsiedniego kanału i interferencja współkanałowa

### Treść zadania

Jaka jest różnica pomiędzy interferencją sąsiedniego kanału (adjacent channel interference) a interferencją współkanałową?

### Odpowiedź

W sieciach komórkowych występują różne rodzaje zakłóceń radiowych. Do najważniejszych należą interferencja sąsiedniego kanału oraz interferencja współkanałowa.

### 1. Interferencja sąsiedniego kanału (Adjacent Channel Interference)

Interferencja sąsiedniego kanału występuje wtedy, gdy sygnały z kanałów o zbliżonych częstotliwościach wzajemnie się zakłócają.

Główną przyczyną tego zjawiska jest niedoskonałość filtrów odbiorczych i nadajników, a także niewystarczające odstępy pomiędzy sąsiednimi kanałami.

**Sposoby ograniczania:**
- Stosowanie odpowiednich filtrów częstotliwościowych.
- Zachowanie odpowiednich odstępów pomiędzy kanałami.
- Odpowiednie planowanie przydziału częstotliwości.

### 2. Interferencja współkanałowa (Co-channel Interference)

Interferencja współkanałowa występuje pomiędzy komórkami wykorzystującymi tę samą częstotliwość.

Zjawisko to jest związane z ponownym wykorzystaniem tych samych częstotliwości w różnych komórkach sieci. Zakłócenia pojawiają się, gdy sygnały pochodzące z innych stacji bazowych wykorzystujących tę samą częstotliwość docierają do odbiornika.

**Sposoby ograniczania:**
- Zwiększenie odległości pomiędzy komórkami wykorzystującymi tę samą częstotliwość.
- Odpowiedni dobór rozmiaru klastra.
- Prawidłowe rozmieszczenie stacji bazowych.
- Kontrola mocy nadawania.

### 3. Najważniejsze różnice

| Cecha | Interferencja sąsiedniego kanału | Interferencja współkanałowa |
|---|---|---|
| Częstotliwości | Sąsiednie częstotliwości | Ta sama częstotliwość |
| Główna przyczyna | Niedoskonałość filtrów i zbyt mały odstęp kanałów | Ponowne wykorzystanie częstotliwości |
| Sposób ograniczania | Filtry i odstępy częstotliwościowe | Zwiększenie odległości między komórkami |

### Wnioski

Interferencja sąsiedniego kanału i interferencja współkanałowa to dwa różne rodzaje zakłóceń występujących w sieciach komórkowych.

Podstawowa różnica polega na tym, że interferencja sąsiedniego kanału wynika z zakłócania się sygnałów o zbliżonych częstotliwościach, natomiast interferencja współkanałowa występuje pomiędzy komórkami korzystającymi z tej samej częstotliwości.

Odpowiednie planowanie częstotliwości, stosowanie filtrów oraz właściwe rozmieszczenie stacji bazowych pozwalają ograniczyć oba rodzaje zakłóceń.

## Zadanie 11 – Zalety podziału komórki na sektory

### Treść zadania

Jaka jest zaleta podziału komórki na sektory?

### Odpowiedź

Podział komórki na sektory (sectorization) polega na podzieleniu obszaru obsługiwanego przez jedną stację bazową na kilka mniejszych sektorów. Najczęściej stosuje się podział na trzy sektory, z których każdy obejmuje kąt 120°.

Każdy sektor wykorzystuje osobną antenę kierunkową, która nadaje i odbiera sygnały w określonym kierunku.

### Główne zalety podziału komórki na sektory:

- **Zmniejszenie interferencji współkanałowej** – anteny kierunkowe ograniczają odbieranie sygnałów zakłócających z innych kierunków.
- **Zwiększenie pojemności sieci** – ograniczenie zakłóceń pozwala efektywniej wykorzystywać dostępne częstotliwości.
- **Poprawa jakości sygnału** – zmniejszenie zakłóceń wpływa na lepszą jakość połączeń.
- **Lepsze wykorzystanie zasobów częstotliwościowych** – umożliwia bardziej efektywne planowanie i wykorzystanie dostępnych kanałów.

### Wnioski

Główną zaletą podziału komórki na sektory jest zmniejszenie interferencji współkanałowej oraz zwiększenie pojemności sieci komórkowej. Dzięki zastosowaniu anten kierunkowych można poprawić jakość sygnału i obsłużyć większą liczbę użytkowników bez konieczności zwiększania liczby stacji bazowych.

### Treść zadania

## Rozwiązanie

### 1. Obliczenie długości fali

Korzystamy ze wzoru:

$$
\lambda = \frac{c}{f}
$$

gdzie:

- $c = 2{,}998 \cdot 10^8$ m/s – prędkość światła,
- $f = 20 \cdot 10^9$ Hz – częstotliwość sygnału.

$$
\lambda = \frac{2{,}998 \cdot 10^8}{20 \cdot 10^9}
$$

$$
\boxed{\lambda \approx 0{,}01499\ m}
$$

### 2. Obliczenie zysku anteny odbiorczej

Przyjmujemy sprawność anteny równą 30%.

Korzystamy ze wzoru:

$$
G_r = \eta \cdot \left(\frac{\pi \cdot D}{\lambda}\right)^2
$$

gdzie:

- $\eta = 0{,}3$ – sprawność anteny,
- $D = 2{,}5$ m – średnica anteny,
- $\lambda = 0{,}01499$ m – długość fali.

Podstawiamy wartości:

$$
G_r = 0{,}3 \cdot
\left(\frac{\pi \cdot 2{,}5}{0{,}01499}\right)^2
$$

$$
G_r \approx 82356
$$

Przeliczamy zysk na decybele:

$$
G_r(dB) = 10 \cdot \log_{10}(82356)
$$

$$
\boxed{G_r \approx 49{,}16\ dB}
$$

### 3. Obliczenie mocy odbieranego sygnału

Korzystamy ze wzoru Friisa:

$$
P_r = P_t \cdot G_r \cdot G_t \cdot
\left(\frac{\lambda}{4\pi d}\right)^2

Najpierw przeliczamy wartości na jednostki liniowe:

- $P_t = 30$ mW = $0{,}03$ W,
- $G_t = 10^{30/10} = 1000$,
- $G_r = 82356$,

Podstawiamy wartości do wzoru:

$$
P_r = 0{,}03 \cdot 82356 \cdot 1000
\cdot \left(\frac{0{,}01499}{4\pi \cdot 5000}\right)^2
$$

$$
P_r \approx 1{,}407 \cdot 10^{-7}\ W
$$

Po przeliczeniu na mikrowaty:

$$
P_r = 1{,}407 \cdot 10^{-7} \cdot 10^6
$$

$$
\boxed{P_r \approx 0{,}141\ \mu W}
$$

# Zadanie 4 – Wielodrogowość i współczynnik przekraczania progu

### Treść zadania

W sieci komórkowej sygnał dociera do odbiornika w różnym czasie z powodu odbicia, rozproszenia i załamania. Wyjaśnij, w jaki sposób można rozróżnić i przetwarzać przychodzące sygnały. Oblicz współczynnik przekraczania progu zakładając, że wartość progu wynosi 1, a średnia kwadratowa amplitudy wynosi 1,02. Odbiornik porusza się z prędkością 20 km/h, częstotliwość sygnału wynosi 800 MHz.

## Rozwiązanie

### 1. Wyjaśnienie zjawiska wielodrogowości

Wielodrogowość (multipath propagation) polega na docieraniu sygnału radiowego do odbiornika różnymi drogami.

Sygnał może ulegać:
- odbiciu od budynków i innych przeszkód,
- rozproszeniu na różnych obiektach,
- załamaniu podczas przechodzenia przez różne ośrodki.

W wyniku tego do odbiornika docierają kopie tego samego sygnału z różnymi opóźnieniami, amplitudami i fazami.

Do rozróżniania i przetwarzania przychodzących sygnałów można zastosować technikę RAKE, która wykorzystuje kilka odbiorników (palców RAKE) do odbierania różnych kopii sygnału. Następnie sygnały są odpowiednio synchronizowane i łączone, co pozwala poprawić jakość odbioru.

### 2. Obliczenie maksymalnej częstotliwości Dopplera

Najpierw przeliczamy prędkość odbiornika na metry na sekundę:

$$
v = \frac{20 \cdot 1000}{3600}
$$

$$
v \approx 5{,}56\ m/s
$$

Obliczamy długość fali:

$$
\lambda = \frac{c}{f}
$$

gdzie:

- $c = 3 \cdot 10^8$ m/s – prędkość światła,
- $f = 800 \cdot 10^6$ Hz – częstotliwość sygnału.

Podstawiamy wartości:

$$
\lambda = \frac{3 \cdot 10^8}{800 \cdot 10^6}
$$

$$
\boxed{\lambda = 0{,}375\ m}
$$

Częstotliwość Dopplera obliczamy ze wzoru:

$$
f_D = \frac{v}{\lambda}
$$

Podstawiamy wartości:

$$
f_D = \frac{5{,}56}{0{,}375}
$$

$$
\boxed{f_D \approx 14{,}81\ Hz}
$$

### 3. Obliczenie współczynnika przekraczania progu

Korzystamy ze wzoru na współczynnik przekraczania progu (Level Crossing Rate):

$$
N_R = \sqrt{2\pi} \cdot f_D \cdot
\frac{R}{R_{rms}} \cdot
e^{-\frac{R^2}{2R_{rms}^2}}
$$

gdzie:

- $R = 1$ – wartość progu,
- $R_{rms} = 1{,}02$ – średnia kwadratowa amplitudy,
- $f_D = 14{,}81$ Hz – maksymalna częstotliwość Dopplera.

Podstawiamy wartości:

$$
N_R = \sqrt{2\pi} \cdot 14{,}81 \cdot
\frac{1}{1{,}02} \cdot
e^{-\frac{1^2}{2 \cdot 1{,}02^2}}
$$

Obliczamy:

$$
N_R \approx 22{,}51
$$

$$
\boxed{N_R \approx 22{,}51\ przekroczeń/s}
$$

## Wnioski

Zjawisko wielodrogowości powoduje, że do odbiornika docierają kopie sygnału z różnymi opóźnieniami, amplitudami i fazami. Do ich rozróżniania i łączenia można wykorzystać technikę RAKE.

Maksymalna częstotliwość Dopplera wynosi około 14,81 Hz. Obliczony współczynnik przekraczania progu wynosi około 22,51 przekroczeń na sekundę.

Oznacza to, że amplituda odbieranego sygnału przekracza określony próg średnio około 22,51 razy w ciągu sekundy.

## Zadanie 7 – Propagacja sygnału w przestrzeni bez przeszkód i pomiędzy przeszkodami

### 1. Propagacja sygnału w wolnej przestrzeni

Propagacja w wolnej przestrzeni oznacza rozchodzenie się fal radiowych bez przeszkód pomiędzy nadajnikiem a odbiornikiem.

Charakterystyczne cechy:

- Sygnał dociera bezpośrednio od nadajnika do odbiornika.
- Nie występują odbicia ani rozproszenia sygnału.
- Moc sygnału maleje wraz ze wzrostem odległości.
- Propagacja jest bardziej przewidywalna.

### 2. Propagacja sygnału pomiędzy przeszkodami

W rzeczywistym środowisku fale radiowe napotykają różne przeszkody, takie jak budynki, drzewa i inne obiekty.

Występują wtedy następujące zjawiska:

- Odbicie (reflection) – fala odbija się od powierzchni przeszkód.
- Dyfrakcja (diffraction) – fala ugina się na krawędziach przeszkód.
- Rozproszenie (scattering) – fala zmienia kierunek na skutek kontaktu z nierównymi powierzchniami.
- Pochłanianie (absorption) – część energii sygnału zostaje pochłonięta przez przeszkody.

W rezultacie sygnał może docierać do odbiornika różnymi drogami i w różnym czasie.

### 3. Najważniejsze różnice

| Cecha | Wolna przestrzeń | Pomiędzy przeszkodami |
|---|---|---|
| Droga sygnału | Bezpośrednia | Wiele różnych dróg |
| Odbicia | Brak | Występują |
| Rozproszenie | Brak | Występuje |
| Tłumienie | Przewidywalne | Zmiennie i często większe |
| Jakość sygnału | Bardziej stabilna | Może ulegać zmianom |

## Wnioski

Propagacja sygnału w wolnej przestrzeni jest prostsza i bardziej przewidywalna, ponieważ fala dociera bezpośrednio do odbiornika.

Natomiast propagacja pomiędzy przeszkodami powoduje odbicia, dyfrakcję, rozproszenie i pochłanianie fal radiowych. Zjawiska te mogą powodować większe tłumienie, opóźnienia oraz zmiany jakości odbieranego sygnału.



## Zadanie 8 – Różnica pomiędzy szybkim i wolnym tłumieniem sygnału

### 1. Szybkie tłumienie sygnału (Fast Fading)

Szybkie tłumienie sygnału polega na gwałtownych zmianach amplitudy i mocy odbieranego sygnału w krótkim czasie.

Główne przyczyny:

- Odbieranie wielu kopii sygnału docierających różnymi drogami.
- Ruch odbiornika lub nadajnika.
- Efekt Dopplera.
- Interferencja sygnałów docierających z różnych kierunków.

Charakterystyczne cechy:

- Zmiany sygnału zachodzą bardzo szybko.
- Występują na niewielkich odległościach.
- Mogą powodować chwilowe zaniki sygnału.

### 2. Wolne tłumienie sygnału (Slow Fading)

Wolne tłumienie sygnału polega na stopniowych zmianach średniej mocy odbieranego sygnału na większych odległościach lub w dłuższym czasie.

Główne przyczyny:

- Przemieszczanie się odbiornika w różnych obszarach.
- Zmiana odległości od stacji bazowej.
- Obecność dużych przeszkód, takich jak budynki i wzgórza.
- Zmiany warunków propagacji fal radiowych.

Charakterystyczne cechy:

- Zmiany sygnału zachodzą stopniowo.
- Występują na większych odległościach.
- Mogą powodować długotrwałe osłabienie sygnału.

### 3. Najważniejsze różnice

| Cecha | Fast Fading | Slow Fading |
|---|---|---|
| Szybkość zmian | Bardzo szybka | Powolna |
| Przyczyna | Wielodrogowość i efekt Dopplera | Przeszkody i zmiany otoczenia |
| Odległość | Niewielka | Większa |
| Charakter zmian | Gwałtowne wahania | Stopniowe zmiany |

## Wnioski

Szybkie tłumienie sygnału powoduje gwałtowne zmiany amplitudy sygnału w krótkim czasie, głównie na skutek wielodrogowości i efektu Dopplera.

Wolne tłumienie sygnału powoduje stopniowe zmiany średniej mocy sygnału, najczęściej na skutek przeszkód terenowych i zmian otoczenia.

Oba zjawiska wpływają na jakość transmisji radiowej w sieciach komórkowych.


# Zadanie 9 – Odbiór sygnału poza polem widzenia stacji bazowej

## Treść zadania

Odbiornik nie znajduje się w polu widzenia stacji bazowej. Wyjaśnij, dlaczego mimo to odbiera sygnał.

## Odpowiedź

Brak bezpośredniej widoczności między stacją bazową a odbiornikiem nie oznacza braku możliwości odbioru sygnału radiowego.

Fale radiowe mogą docierać do odbiornika dzięki kilku zjawiskom propagacyjnym:

### 1. Dyfrakcja (ugięcie fal)
Fale radiowe mogą uginać się na krawędziach budynków, wzgórz i innych przeszkód, dzięki czemu docierają do obszarów znajdujących się poza bezpośrednim polem widzenia stacji bazowej.

### 2. Odbicie fal
Sygnał może odbijać się od budynków, ścian i innych powierzchni, zmieniając kierunek propagacji i docierając do odbiornika inną drogą.

### 3. Rozpraszanie fal
Fale radiowe mogą rozpraszać się na różnych obiektach, takich jak drzewa, budynki czy inne przeszkody. Część energii sygnału dociera w ten sposób do odbiornika.

### 4. Przenikanie przez przeszkody
Część energii fal radiowych może przenikać przez ściany, okna i inne materiały budowlane, umożliwiając odbiór sygnału wewnątrz budynków.

## Wnioski

Odbiornik może odbierać sygnał nawet wtedy, gdy nie znajduje się w bezpośrednim polu widzenia stacji bazowej. Jest to możliwe dzięki zjawiskom dyfrakcji, odbicia, rozpraszania oraz przenikania fal radiowych przez przeszkody.

Sygnał docierający do odbiornika może mieć jednak mniejszą moc i być bardziej podatny na zaniki oraz zakłócenia.

# Zadanie 10 – Efekt Dopplera i przesunięcie częstotliwości

## Treść zadania

Antena transmituje sygnał o częstotliwości 900 MHz. Odbiornik porusza się z prędkością 40 km/h. Oblicz częstotliwość (przesunięcie) Dopplera.

## Odpowiedź

Efekt Dopplera polega na zmianie częstotliwości odbieranego sygnału w wyniku ruchu odbiornika względem nadajnika.

Przesunięcie częstotliwości Dopplera obliczamy ze wzoru:

\[
f_D = \frac{v}{c} \cdot f
\]

gdzie:

- \(f_D\) – przesunięcie częstotliwości Dopplera,
- \(v\) – prędkość odbiornika,
- \(c\) – prędkość światła (300 000 000 m/s),
- \(f\) – częstotliwość nadawanego sygnału.

### 1. Przeliczenie prędkości odbiornika

Prędkość odbiornika wynosi 40 km/h.

Przeliczamy ją na m/s:

\[
v = \frac{40 \cdot 1000}{3600}
\]

\[
v \approx 11,11 \text{ m/s}
\]

### 2. Obliczenie przesunięcia Dopplera

Podstawiamy wartości do wzoru:

\[
f_D = \frac{11,11}{300\,000\,000} \cdot 900\,000\,000
\]

\[
f_D \approx 33,33 \text{ Hz}
\]

<math block value="\\boxed{f_D \\approx 33,33\\text{ Hz}}"/>

## Wnioski

Przesunięcie częstotliwości Dopplera wynosi około 33,33 Hz.

Jeżeli odbiornik porusza się w kierunku nadajnika, częstotliwość odbieranego sygnału wzrasta. Natomiast podczas oddalania się od nadajnika częstotliwość maleje.

Wartość przesunięcia zależy od prędkości odbiornika oraz częstotliwości nadawanego sygnału.
